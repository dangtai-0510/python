```python
import math


def nhap_so(ten):
    x = int(input("Nhap " + ten + ": "))
    while x <= 0:
        print("Phai nhap so nguyen duong!")
        x = int(input("Nhap lai " + ten + ": "))
    return x


def la_tam_giac(a, b, c):
    return a + b > c and a + c > b and b + c > a


def chu_vi(a, b, c):
    return a + b + c


def dien_tich(a, b, c):
    p = chu_vi(a, b, c) / 2
    return math.sqrt(p * (p - a) * (p - b) * (p - c))


a = nhap_so("a")
b = nhap_so("b")
c = nhap_so("c")

if la_tam_giac(a, b, c):
    print("Day la 3 canh cua mot tam giac.")
    print("Chu vi:", chu_vi(a, b, c))
    print("Dien tich:", round(dien_tich(a, b, c), 2))
else:
    print("Day khong phai la 3 canh cua mot tam giac.")
```
