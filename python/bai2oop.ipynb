```python
don_hang = [
    ("O01", "HN", 1200000, "done"),
    ("O02", "HCM", 850000, "pending"),
    ("O03", "HN", 3500000, "done"),
    ("O04", "DN", 0, "cancelled"),
    ("O05", "HCM", 2100000, "done"),
    ("O06", "HN", 670000, "pending")
]

# a
xong = list(filter(lambda i: i[3] == "done", don_hang))

print("a)")
for i in xong:
    print(i)

# b
sap_xep = sorted(don_hang, key=lambda i: i[2])

print("\nb)")
for i in sap_xep:
    print(i)

# c
vat = list(map(lambda i: (i[0], i[2] * 1.1), xong))

print("\nc)")
for i in vat:
    print(i)

# d
lon = max(don_hang, key=lambda i: i[2])
nho = min(don_hang, key=lambda i: i[2])

print("\nd)")
print("Lớn nhất:", lon)
print("Nhỏ nhất:", nho)
```
