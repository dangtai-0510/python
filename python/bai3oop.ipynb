```python
san_pham = [
    ("SP01", 50000, 300),
    ("SP02", 150000, 100),
    ("SP03", 70000, 0),
    ("SP04", 250000, 60),
    ("SP05", 90000, 150),
    ("SP06", 450000, 40),
    ("SP07", 120000, 20)
]


con_hang = list(filter(lambda i: i[2] > 0, san_pham))


gia_tri = list(map(lambda i: (i[0], i[1], i[2], i[1] * i[2]), con_hang))


tren10 = list(filter(lambda i: i[3] > 10000000, gia_tri))


giam = sorted(tren10, key=lambda i: i[3], reverse=True)


print("Top 5:")

for i in giam[:5]:
    print(i)
```
