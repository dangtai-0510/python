```python
# Bai 2 - Mo phong don gian chuoi cung ung

import random

ton_kho = 1000
ngay = 1
so_lan_dat_hang = 0
doanh_thu = 0
ton_kho_thap_nhat = ton_kho

ngay_nhan_hang = 0

print("BANG THEO DOI TON KHO")
print("-" * 75)
print("%-6s | %-8s | %-10s | %-20s" %
      ("Ngay", "Ban", "Ton kho", "Ghi chu"))
print("-" * 75)

while ngay <= 30:

    ghi_chu = ""

    # Nhan hang neu den ngay
    if ngay == ngay_nhan_hang:
        ton_kho = ton_kho + 500
        ghi_chu = "Nhan 500 SP"

    # Ban ngau nhien tu 50 den 150 SP
    so_luong_ban = random.randint(50, 150)

    # Khong ban qua so luong ton kho
    if so_luong_ban > ton_kho:
        so_luong_ban = ton_kho

    ton_kho = ton_kho - so_luong_ban

    # Tinh doanh thu
    doanh_thu = doanh_thu + so_luong_ban * 150000

    # Tim ton kho thap nhat
    if ton_kho < ton_kho_thap_nhat:
        ton_kho_thap_nhat = ton_kho

    # Dat hang khi ton kho duoi 200
    if ton_kho < 200 and ngay_nhan_hang == 0:
        so_lan_dat_hang = so_lan_dat_hang + 1
        ngay_nhan_hang = ngay + 2

        if ghi_chu == "":
            ghi_chu = "Dat hang 500 SP"
        else:
            ghi_chu = ghi_chu + " | Dat hang 500 SP"

    # In thong tin tung ngay
    print("%-6d | %-8d | %-10d | %-20s" %
          (ngay, so_luong_ban, ton_kho, ghi_chu))

    # Reset sau khi da nhan hang
    if ngay == ngay_nhan_hang:
        ngay_nhan_hang = 0

    ngay = ngay + 1

print("-" * 75)
print("Tong doanh thu:", doanh_thu, "dong")
print("So lan dat hang:", so_lan_dat_hang)
print("Ton kho thap nhat:", ton_kho_thap_nhat, "SP")
```
